Questão 5

```python
number = int(input('Digite um número entre 1000 e 9999: '))

while number < 1000 or number > 9999:
    print('Valor inválido, tente novamente.')
    number = int(input('Digite um número entre 1000 e 9999: '))

thousands = number // 1000
hundreds = (number // 100) % 10
tens = (number // 10) % 10
units = number % 10

print(f'Unidade: {units}')
print(f'Dezena: {tens}')
print(f'Centena: {hundreds}')
print(f'Milhar: {thousands}')
```

Questão 7

```python
text = input('Digite uma palavra qualquer: ')
vowels = ['a', 'e', 'i', 'o', 'u']
vowel_count = 0
has_a = False

for letter in text:
  print(letter)

  if letter in vowels:
    vowel_count += 1

    if letter == vowels[0]:
      has_a = True

print(f'Quantidade de vogais na palavra fornecida: {vowel_count}')

if has_a:
  print(f'A palavra fornecida possui a letra \'a\'.')
else:
  print(f'A palavra fornecida não possui a letra \'a\'.')
```

Questão 8

```python
first_number = int(input('Digite o primeiro número: '))
second_number = int(input('Digite o segundo número: '))

while second_number == 0:
  print(f'Segundo número não pode ser nulo por conta da divisão.')
  second_number = int(input('Digite o segundo número: '))

print(f'Adição: {first_number + second_number}')
print(f'Subtração: {first_number - second_number}')
print(f'Multiplicação: {first_number * second_number}')
print(f'Divisão: {float(first_number) / float(second_number)}')
print(f'Resto da divisão: {first_number % second_number}')
print(f'Potenciação: {first_number ** second_number}')

addition = first_number + second_number

if addition % 2 == 0:
  print(f'Resultado da adição é par')
else:
  print(f'Resultado da adição é ímpar')
```
