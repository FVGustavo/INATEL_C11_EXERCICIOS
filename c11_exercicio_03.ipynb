Questão 4

```python
people = {}

for i in range(3):
    name    = input(f"Nome da pessoa {i + 1}: ")
    weight  = float(input(f"Peso de {name}: "))
    people[name] = weight

heaviest = max(people, key=people.get)
lightest = min(people, key=people.get)

print("Mais pesado:", heaviest)
print("Mais leve:", lightest)
```

Questão 5

```python
n                 = int(input("Quantas pessoas? "))

age_sum           = 0
women_under_20  = 0

for i in range(n):
    name  = input("Nome: ")
    age   = int(input("Idade: "))
    sex   = input("Sexo (M/F): ").strip().upper()

    age_sum += age
    if sex == "F" and age < 20:
        women_under_20 += 1

print("Média de idade:", age_sum / n)
print("Mulheres com menos de 20 anos:", women_under_20)
```

```python
recipe                              = {'ovo', 'farinha', 'açúcar', 'trigo', 'fermento', 'leite', 'manteiga', 'chocolate', 'cenoura'}
first_person_ingredients            = {'ovo', 'chocolate', 'manteiga'}
second_person_ingredients           = {'trigo', 'fermento', 'leite', 'manteiga'}
first_person_missing_ingredients    = recipe - first_person_ingredients
second_person_missing_ingredients   = recipe - second_person_ingredients

print(f'Ingredientes faltantes da primeira pessoa: {first_person_missing_ingredients}')
print(f'Ingredientes faltantes da segunda pessoa: {second_person_missing_ingredients}')
```

Questão 8

```python
product_list  = []

for i in range(3):
  product               = {}
  product["name"]       = input(f"Insira o nome do produto {i + 1}: ")
  product["price"]      = float(input(f"Insira o preço do produto {i + 1}: "))
  product["quantity"]   = int(input(f"Insira a quantidade em estoque do produto {i + 1}: "))

  product_list.append(product)

for product in product_list:
  print(product["name"])
  print(product["price"] * product["quantity"])
```
